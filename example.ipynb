```python
import pandas as pd
```

```python
df =  pd.DataFrame({"x":[1,2,3,4,5],"j":[1,4,9,16,25]})
df
```

# Simple Linear Regression

```python
X=df.drop(columns="j")
y=df["j"]                  #y should be a 1D

#Modelling
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()
lr_model.fit(X,y)

print("coeeffient:",lr_model.coef_)
print("intercept:",lr_model.intercept_)

#Prediction
ypred = lr_model.predict(X)
print("Model predicted values:",ypred)

#Evalaution
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y,ypred)
print("Mean Square Error:", mse)

from sklearn.metrics import r2_score
r2 = r2_score(y, ypred)
print("R2 Score:",r2)
```

# Non Linear or Polynominal Regression

```python
#increasing degree of x
df["x2"] = df["x"]**2
df["x3"] = df["x"]**3

#select i/p & o/p
X=df.drop(columns="j")
y=df["j"]                  #y should be a 1D

#Modelling
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()
lr_model.fit(X,y)

print("coeeffient:",lr_model.coef_)
print("intercept:",round(lr_model.intercept_,2))

#Prediction
ypred = lr_model.predict(X)
print("Model predicted values:",ypred)

#Evalaution
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y,ypred)
print("Mean Square Error:", round(mse,2))

from sklearn.metrics import r2_score
r2 = r2_score(y, ypred)
print("R2 Score:",r2)
```
